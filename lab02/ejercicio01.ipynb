```python
multiplo_3 = lambda x: x % 3 == 0
cubo_num = lambda x : x**3
producto = lambda a, b: a*b
mayor_num = lambda a, b: max(a, b)
inicia_con_a = lambda word: word[0] == "a" or word[0] == "A"
convert_celcius_fah = lambda x: (x*1.8) + 32
```

```python
#pruebas
print(multiplo_3(6))
print(cubo_num(3))
print(producto(5, 2))
print(mayor_num(17, 14))
print(convert_celcius_fah(0))
```
