```python
from functools import reduce

numeros = [3, 5, 12, 16, 23, 41]
palabras = ["mantequilla", "hola", "mundo", "lapiz", "examen", "agua"]

multiplicacion = reduce(lambda x, y: x*y, numeros)
concatenar = reduce(lambda a, b: a + "" + b, palabras)

def mayor(a, b):
    return max(a, b)

mayor_numero = reduce(lambda x, y: mayor(x, y), numeros)
```

```python
#verificacion
print(multiplicacion)
print(concatenar)
print(mayor_numero)
```
