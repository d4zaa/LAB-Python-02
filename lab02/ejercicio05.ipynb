```python
numeros = [3, 2, 5, 4, 6, 8, 13, 14]

pares = filter(lambda n: n%2==0, numeros)
al_cuadrado = sorted(map(lambda n: n**2, pares))

al_cubo = map(lambda n: n**3, numeros)
mayores_a100 = list(filter(lambda n: n>100, al_cubo))

mayores_a10 = list(filter(lambda n: n>10, numeros))
```

```python
#verificacion
print(al_cuadrado)
print(mayores_a100)
```
