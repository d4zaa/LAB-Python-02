```python
numeros = [3, 6, 7, 8, 11, 23]
celcius = [0, 17, 23, 24, 29]
words = ["Escuela", "Colombiana", "De", "Ingenieria", "Piña"]
matriz = [[1,2,3], [4,5,6], [7,8,9]]

cuadrados = [(x, x**2) for x in numeros]
cubos_mayoresA10 = [x**3 for x in numeros if x>10]
pares = [x for x in numeros if x%2==0]
lista_valores = [f"valor: {x}" for x in numeros]
conv_to_fah = [x * 1.8 + 32 for x in celcius]
palabras_cortas = [word for word in words if len(word)<=5]
iniciales = [word[0] for word in words]
par_o_impar = ["par" if x%2==0 else "impar" for x in numeros]
matriz_to_row = [row[x]:row[x+1] for x in matriz]
```

```python
#verificacion
print(cuadrados)
print(cubos_mayoresA10)
print(pares)
print(lista_valores)
print(conv_to_fah)
print(palabras_cortas)
print(iniciales)
print(par_o_impar)
print(matriz_to_row)
```
