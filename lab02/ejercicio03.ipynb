```python
from math import pi
temperaturas = [0, 13, 16, 21, 32]
list1 = [1, 2, 3, 4]
list2 = [5, 6, 7, 8]
radios = [1, 3, 7, 10]
productos = {"pan": 1000, "leche": 2500, "café": 5000}
matriz = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

conv_to_fah = list(map(lambda x: x*1.8 +32, temperaturas))
unir_listas = list(map(lambda a, b: a+b, list1, list2))
areas_circ = list(map(lambda x: pi * x**2, radios))
dicc_to_tuplas = list(map(lambda item: (item[0], item[1] *1.1), productos.items()))
matriz_x10 = list(map(lambda fila: list(map(lambda x: x * 10, fila)), matriz))
```

```python
#verificacion
print(conv_to_fah)
print(unir_listas)
print(areas_circ)
print(dicc_to_tuplas)
print(matriz_x10)
```
