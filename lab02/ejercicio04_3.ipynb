```python
palabras = ["numero", "arcilla", "madera", "filamento","mouse", "suelo", "cavernicola"]
ordenadas = sorted(palabras, key = lambda x: x[0])

tuplas = [(2, "amarillo"), (9, "naranja"), (5, "rojo"), (11, "verde")]
t_ordenadas= sorted(tuplas, key = lambda t: t[0])

productos = [
      ("Camisa", 30),
      ("Pantalón", 55),
      ("Medias", 10),
      ("Chaqueta", 80)
  ]
p_ordenados = sorted(productos, key = lambda p: (-p[1], p[0]))

palabras_por_longitud = sorted(palabras, key = lambda x: len(x))

numeros = [3, 4, 15, 12, 6, 81, 34]
pares_e_impares = sorted(numeros, key = lambda n: (n%3==0, n%2==0))

palabra = "paralelepipedo"
frecuencia = {}

for letra in palabra:
    frecuencia[letra] = frecuencia.get(letra, 0) + 1

frecuencia_ordenada = dict(
    sorted(frecuencia.items(), key=lambda item: item[1], reverse=True)
)


empleados = [
      ("Maria", "Ventas", 30),
      ("Luis", "Tecnología", 25),
      ("Ana", "Ventas", 25),
      ("Pedro", "Tecnología", 28)
    ]
```

```python
#verificacion
print(ordenadas)
print(t_ordenadas)
print(p_ordenados)
print(palabras_por_longitud)
print(pares_e_impares)
print(frecuencia_ordenada)
```
